```python
import numpy as np

key = np.array([[3, 3],
                [2, 5]])

def encrypt(text):
    text = text.upper().replace(" ", "")

    if len(text) % 2 != 0:
        text += 'X'

    cipher = ""

    for i in range(0, len(text), 2):
        P = np.array([ord(text[i]) - 65,
                      ord(text[i+1]) - 65])

        C = np.dot(key, P) % 26

        cipher += chr(C[0] + 65) + chr(C[1] + 65)

    return cipher


def decrypt(cipher):
    det = round(np.linalg.det(key))
    det_inv = pow(det % 26, -1, 26)

    adj = np.array([[ key[1][1], -key[0][1]],
                    [-key[1][0],  key[0][0]]])

    key_inv = (det_inv * adj) % 26

    text = ""

    for i in range(0, len(cipher), 2):
        C = np.array([ord(cipher[i]) - 65,
                      ord(cipher[i+1]) - 65])

        P = np.dot(key_inv, C) % 26

        text += chr(P[0] + 65) + chr(P[1] + 65)

    return text


text = input("Enter plaintext: ")

cipher = encrypt(text)

print("Encrypted:", cipher)
print("Decrypted:", decrypt(cipher))
```
