```python
# OS LAB - File Storage Allocation Techniques
# 1. Contiguous Allocation
# 2. Linked Allocation
# 3. Indexed (Indirect) Allocation


# ---------------------------------------------------------
# INPUT
# ---------------------------------------------------------

file_name = input("Enter file name: ")
start_block = int(input("Enter starting block: "))
num_blocks = int(input("Enter number of blocks: "))


# ---------------------------------------------------------
# 1. CONTIGUOUS ALLOCATION
# ---------------------------------------------------------

def contiguous_allocation(start, n):
    blocks = []

    for i in range(n):
        blocks.append(start + i)

    print("\n--- Contiguous Allocation ---")
    print("File Name:", file_name)
    print("Starting Block:", start)
    print("Allocated Blocks:", blocks)


# ---------------------------------------------------------
# 2. LINKED ALLOCATION
# ---------------------------------------------------------

class Node:
    def __init__(self, block):
        self.block = block
        self.next = None


def linked_allocation(start, n):
    head = None
    current = None

    # Create linked list
    for i in range(n):
        node = Node(start + i * 2)

        if head is None:
            head = node
            current = node
        else:
            current.next = node
            current = node

    # Display linked list
    print("\n--- Linked Allocation ---")
    print("File Name:", file_name)

    current = head

    while current:
        print(current.block, end=" -> ")
        current = current.next

    print("NULL")


# ---------------------------------------------------------
# 3. INDEXED / INDIRECT ALLOCATION
# ---------------------------------------------------------

def indexed_allocation(start, n):
    index_block = start
    data_blocks = []

    for i in range(n):
        data_blocks.append(start + i + 1)

    print("\n--- Indexed (Indirect) Allocation ---")
    print("File Name:", file_name)
    print("Index Block:", index_block)
    print("Data Blocks:", data_blocks)

    print("\nIndex Block")
    print("    |")

    for block in data_blocks:
        print("    +---->", block)


# ---------------------------------------------------------
# MAIN PROGRAM
# ---------------------------------------------------------

contiguous_allocation(start_block, num_blocks)

linked_allocation(start_block, num_blocks)

indexed_allocation(start_block, num_blocks)
```
