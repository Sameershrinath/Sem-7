## AIM

To implement and study various CPU scheduling algorithms:

a) First Come First Serve (FCFS) <br>
b) Shortest Job First (SJF)<br>
c) Round Robin (RR)<br>
d) Priority Scheduling<br>

```python
# ==========================================
# CPU SCHEDULING ALGORITHMS
# FCFS, SJF, ROUND ROBIN, PRIORITY
# ==========================================


# ------------------------------------------
# INPUT SECTION - Input is taken only once
# ------------------------------------------

n = int(input("Enter number of processes: "))

processes = []

for i in range(n):
    pid = "P" + str(i + 1)

    at = int(input(f"Enter Arrival Time for {pid}: "))
    bt = int(input(f"Enter Burst Time for {pid}: "))
    priority = int(input(f"Enter Priority for {pid}: "))

    processes.append([pid, at, bt, priority])

# Time Quantum is required for Round Robin
tq = int(input("Enter Time Quantum: "))


# ==========================================
# 1. FCFS - FIRST COME FIRST SERVE
# ==========================================

def fcfs(processes):

    # Sort processes according to Arrival Time
    p = sorted(processes, key=lambda x: x[1])

    time = 0
    total_wt = 0
    total_tat = 0

    print("\n----- FCFS Scheduling -----")
    print("PID\tAT\tBT\tCT\tTAT\tWT")

    for pid, at, bt, priority in p:

        # If CPU is idle
        if time < at:
            time = at

        # Completion Time
        time += bt
        ct = time

        # Turnaround Time = CT - AT
        tat = ct - at

        # Waiting Time = TAT - BT
        wt = tat - bt

        total_wt += wt
        total_tat += tat

        print(f"{pid}\t{at}\t{bt}\t{ct}\t{tat}\t{wt}")

    print("Average WT =", total_wt / len(p))
    print("Average TAT =", total_tat / len(p))


# ==========================================
# 2. SJF - SHORTEST JOB FIRST
# Non-Preemptive
# ==========================================

def sjf(processes):

    # Make a copy of process list
    p = processes.copy()

    n = len(p)

    completed = [False] * n

    time = 0
    count = 0

    total_wt = 0
    total_tat = 0

    print("\n----- SJF Scheduling -----")
    print("PID\tAT\tBT\tCT\tTAT\tWT")

    while count < n:

        idx = -1
        min_bt = float('inf')

        # Find arrived process having minimum Burst Time
        for i in range(n):

            if p[i][1] <= time and not completed[i]:

                if p[i][2] < min_bt:
                    min_bt = p[i][2]
                    idx = i

        # If no process has arrived, CPU remains idle
        if idx == -1:
            time += 1
            continue

        pid, at, bt, priority = p[idx]

        # Completion Time
        time += bt
        ct = time

        # Turnaround Time
        tat = ct - at

        # Waiting Time
        wt = tat - bt

        total_wt += wt
        total_tat += tat

        print(f"{pid}\t{at}\t{bt}\t{ct}\t{tat}\t{wt}")

        completed[idx] = True
        count += 1

    print("Average WT =", total_wt / n)
    print("Average TAT =", total_tat / n)


# ==========================================
# 3. ROUND ROBIN
# ==========================================

def round_robin(processes, tq):

    from collections import deque

    # Sort according to Arrival Time
    p = sorted(processes, key=lambda x: x[1])

    n = len(p)

    # Store remaining Burst Time
    remaining = [x[2] for x in p]

    # Store Completion Time
    ct = [0] * n

    queue = deque()

    time = 0
    i = 0

    while i < n or queue:

        # If CPU is idle, move time to next process
        if not queue and i < n and time < p[i][1]:
            time = p[i][1]

        # Add arrived processes into Ready Queue
        while i < n and p[i][1] <= time:
            queue.append(i)
            i += 1

        if not queue:
            continue

        # Get first process from queue
        idx = queue.popleft()

        # Process executes for Time Quantum
        run = min(tq, remaining[idx])

        time += run

        remaining[idx] -= run

        # Add newly arrived processes
        while i < n and p[i][1] <= time:
            queue.append(i)
            i += 1

        # If process is not completed,
        # put it back into Ready Queue
        if remaining[idx] > 0:
            queue.append(idx)

        else:
            # Store Completion Time
            ct[idx] = time

    total_wt = 0
    total_tat = 0

    print("\n----- Round Robin Scheduling -----")
    print("PID\tAT\tBT\tCT\tTAT\tWT")

    for i in range(n):

        pid, at, bt, priority = p[i]

        # Turnaround Time
        tat = ct[i] - at

        # Waiting Time
        wt = tat - bt

        total_wt += wt
        total_tat += tat

        print(f"{pid}\t{at}\t{bt}\t{ct[i]}\t{tat}\t{wt}")

    print("Average WT =", total_wt / n)
    print("Average TAT =", total_tat / n)


# ==========================================
# 4. PRIORITY SCHEDULING
# Non-Preemptive
# Smaller number = Higher Priority
# ==========================================

def priority_scheduling(processes):

    p = processes.copy()

    n = len(p)

    completed = [False] * n

    time = 0
    count = 0

    total_wt = 0
    total_tat = 0

    print("\n----- Priority Scheduling -----")
    print("PID\tAT\tBT\tPR\tCT\tTAT\tWT")

    while count < n:

        idx = -1
        highest_priority = float('inf')

        # Find highest priority process
        # among arrived processes
        for i in range(n):

            if p[i][1] <= time and not completed[i]:

                if p[i][3] < highest_priority:
                    highest_priority = p[i][3]
                    idx = i

        # If no process has arrived
        if idx == -1:
            time += 1
            continue

        pid, at, bt, priority = p[idx]

        # Completion Time
        time += bt
        ct = time

        # Turnaround Time
        tat = ct - at

        # Waiting Time
        wt = tat - bt

        total_wt += wt
        total_tat += tat

        print(
            f"{pid}\t{at}\t{bt}\t{priority}\t"
            f"{ct}\t{tat}\t{wt}"
        )

        completed[idx] = True
        count += 1

    print("Average WT =", total_wt / n)
    print("Average TAT =", total_tat / n)


# ==========================================
# CALL ALL SCHEDULING ALGORITHMS
# ==========================================

fcfs(processes)

sjf(processes)

round_robin(processes, tq)

priority_scheduling(processes)
```
